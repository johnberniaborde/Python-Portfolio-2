# Portfolio 2: Marketing Sales Analyzer
This program helps the department track if our sales cover our costs.
It uses **Conditionals** (If/Else) to give us a status report.

```python
# Asking for inputs
raw_sales = input("Enter total sales (PHP): ")
raw_costs = input("Enter total costs (PHP): ")

try:
    # Converting the text input into numbers (floats)
    sales = float(raw_sales)
    costs = float(raw_costs)
    print("Data received successfully!")
except:
    # If the user types something that isn't a number
    print("Error: Please enter numbers only.")
    sales = 0
    costs = 0
```

```python
# Calculate Profit
profit = sales - costs

# The Decision Machine
if profit > 1000:
    status = "Great Success!"
    suggestion = "Consider buying more inventory."
elif profit > 0:
    status = "Small Profit"
    suggestion = "Keep going, you're doing okay."
elif profit == 0:
    status = "Break Even"
    suggestion = "You didn't lose money, but you didn't make any either."
else:
    status = "Financial Loss"
    suggestion = "We need to cut costs or increase prices."
```

```python
print("---------- REPORT ----------")
print(f"Total Sales:  PHP {sales}")
print(f"Total Costs:  PHP {costs}")
print(f"Net Profit:   PHP {profit}")
print("----------------------------")
print(f"FINAL STATUS: {status}")
print(f"ADVICE:       {suggestion}")
```
